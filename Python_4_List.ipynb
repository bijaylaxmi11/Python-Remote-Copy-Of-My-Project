2. Find the largest and smallest number in a list without using max()/min().

```python
l = [4,6,2,1,9,3]
print(len(l))
for i in range(0,len(l)-1):
    for j in range(i+1,len(l)):
     if l[i] < l[j]:
        l[i],l[j] = l[j],l[i]
print(l)
print(f"the maximum number in list : {l[0]}")
print(f"the minimum number in list : {l[-1]}")
```

# find maximum and minimum in a list using max(),min() 

```python
l = [5,6,3,9]
print(max(l))
print(min(l))
```

nums = [-8, 3, -5, 2]
print(max(nums))          # 3   (largest number as-is)
print(max(nums, key=abs))  # -8  (largest by distance from zero)

3 . Remove the duplicate values from list

```python
l = [3,1,3,4,5,6,4,6,9,8]
li = []
for i in l :
    if i not in li:

        li.append(i)
print(li)
```
