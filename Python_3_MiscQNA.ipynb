WAP to calculate the area and perimeter of rectangle ,triangle

```python
def cal_rectangle(l,h) :
    area = l*h
    print(area)
    perimeter = 2*(l+h)
    print(perimeter)

def cal_triangle(l,b,h) :
    area = 1/2*(b*h)
    print(area)
    perimeter = l+b+h
    print(perimeter)

print(f"rectangle area and perimeter is {cal_rectangle(2,3)})
print(f"triangle area and perimeter is {cal_triangle(2,4,6)})
```
